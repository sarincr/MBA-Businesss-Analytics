## Printing Output

`print()` displays output on the screen.

```
print("Hello World")

print(100)

print(10 + 20)
```

## Variables

Variables store data.

```
name = "Python"

age = 25

price = 99.5

print(name)

print(age)

print(price)
```

## User Input

`input()` takes data from the user.

All input values are stored as **strings** by default.


```
name = input("Enter your name: ")

print(name)

print(type(name))
```

## Why int() Conversion Is Needed

Even if the user enters numbers, `input()` stores them as strings.

Without conversion:
- `"10" + "20"` becomes `"1020"`
- Python joins the text

After conversion using `int()`:
- `10 + 20` becomes `30`
- Python performs mathematical addition


```
a = input("Enter first number: ")

b = input("Enter second number: ")

print(a + b)

print(type(a))

x = int(input("Enter first number again: "))

y = int(input("Enter second number again: "))

print(x + y)

print(type(x))
```

## Integer Input

`int()` converts string input into integers.

```
age = int(input("Enter age: "))

print(age)

print(age + 5)
```

## Float Input

`float()` converts input into decimal numbers.

```
price = float(input("Enter price: "))

print(price)

print(price + 10.5)
```

## Datatype Checking

`type()` checks the datatype.

```
name = "AI"

number = 50

marks = 99.5

print(type(name))

print(type(number))

print(type(marks))
```

## String Concatenation

Strings can be joined using `+`.

```
first = "Machine"

second = "Learning"

result = first + " " + second

print(result)
```

## Arithmetic Operations

Basic mathematical operations.

```
a = int(input("Enter first number: "))

b = int(input("Enter second number: "))

print(a + b)

print(a - b)

print(a * b)

print(a / b)
```

## f-String Formatting

f-Strings create readable output.

```
name = input("Enter name: ")

course = input("Enter course: ")

print(f"Student Name: {name}")

print(f"Selected Course: {course}")
```

## Uppercase and Lowercase

String methods can change text format.

```
text = input("Enter text: ")

print(text.upper())

print(text.lower())
```

## Removing Spaces

`strip()` removes extra spaces.

```
text = input("Enter text with spaces: ")

print(text.strip())
```

## Replacing Text

`replace()` changes words inside a string.

```
text = input("Enter a sentence: ")

print(text.replace("Python", "Java"))
```

## Length of Input

`len()` counts characters.

```
text = input("Enter text: ")

print(len(text))
```

## Multiple Print Methods

Different print formatting methods.

```
print("Python", "Java", "C++")

print("2026", "05", "15", sep="-")

print("Hello", end=" ")

print("World")
```

## Escape Characters

Special formatting characters.

```
print("Hello\nWorld")

print("Python\tProgramming")

print("Welcome \"Students\"")
```

## Boolean Values

Boolean values are `True` or `False`.

```
status = True

result = False

print(status)

print(result)
```

## Simple Percentage Calculation

Using input values in formulas.

```
marks = float(input("Enter marks: "))

total = float(input("Enter total marks: "))

percentage = (marks / total) * 100

print(percentage)
```
